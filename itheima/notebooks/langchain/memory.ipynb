```python
# 短期记忆 short-term memory 持久化
from langgraph.checkpoint.sqlite import SqliteSaver

from langchain.agents import create_agent
with SqliteSaver.from_conn_string("checkpoints.db") as check_pointer:
    agent = create_agent(
        model="deepseek-chat",
        checkpointer=check_pointer
    )

    resp = agent.invoke(
        {
            "messages": [
                HumanMessage("我是米饭好好吃，我喜欢小金毛")
            ]
        },
        {"configurable": {"thread_id": "2"}}
    )
    print(resp)

```

```python
# 记忆管理策略

from langchain.agents import create_agent

from langgraph.checkpoint.sqlite import SqliteSaver

from langchain.agents.middleware import SummarizationMiddleware

with SqliteSaver.from_conn_string("checkpoints.db") as check_pointer:
    agent = create_agent(
        model="deepseek-chat",
        checkpointer=check_pointer,
        middleware = [
            SummarizationMiddleware(
                model="deepseek-chat",
                trigger=("messages", 1),
                keep=("messages", 1)
            )
        ]
    )

    resp = agent.invoke(
        {
            "messages": [
                HumanMessage("我是米饭好好吃")
            ]
        },
        {"configurable": {"thread_id": "3"}}
    )
    resp = agent.invoke(
        {
            "messages": [
                HumanMessage("你好，我是谁？")
            ]
        },
        {"configurable": {"thread_id": "3"}}
    )
    print(resp)
```
