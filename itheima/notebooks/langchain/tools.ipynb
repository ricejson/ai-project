```python
from dotenv import load_dotenv
load_dotenv()

# 工具定义方式1: 使用@tool注解
from langchain.tools import tool
from langchain_core.messages import HumanMessage

@tool("add", description="calculate the sum of two numbers")
def add(num1: int, num2: int) -> int:
    return num1 + num2

from langchain.agents import create_agent
agent = create_agent(
    model="deepseek-chat",
    tools=[add]
)

resp = agent.invoke({
    "messages": [
        HumanMessage("计算2+3的结果")
    ]
})

for message in resp["messages"]:
    message.pretty_print()


```

```python
from dotenv import load_dotenv
load_dotenv()

# 工具定义方式2: 使用注释
from langchain.tools import tool
@tool()
def add(num1: int, num2: int) -> int:
    """
        calculate the sum of two numbers
        Args:
            num1: the first num
            num2: the second num
    """
    return num1 + num2

from langchain.agents import create_agent
agent = create_agent(
    model="deepseek-chat",
    tools=[add]
)

from langchain_core.messages import HumanMessage
resp = agent.invoke({
    "messages": [
        HumanMessage("2+2结果是多少？")
    ]
})
for message in resp["messages"]:
    message.pretty_print()
```

```python
from dotenv import load_dotenv
load_dotenv()

# 工具定义方式3: 使用pydantic + args_schema
from langchain.tools import tool
from pydantic import BaseModel, Field

class Wrapper(BaseModel):
    num1: int = Field(description="被加数")
    num2: int = Field(description="加数")

@tool(args_schema=Wrapper)
def add(wrapper: Wrapper) -> int:
    """
        calculate the sum of two numbers
        Args:
            num1: the first num
            num2: the second num
    """
    return wrapper.num1 + wrapper.num2

from langchain.agents import create_agent
agent = create_agent(
    model="deepseek-chat",
    tools=[add]
)

from langchain_core.messages import HumanMessage
resp = agent.invoke({
    "messages": [
        HumanMessage("2+2结果是多少？")
    ]
})
for message in resp["messages"]:
    message.pretty_print()
```

```python
# 读取环境变量
from dotenv import load_dotenv
load_dotenv()

# tavily_search tool
from langchain_tavily import TavilySearch
search_tool = TavilySearch(
    max_results=2,
)

# 定义agent
from langchain.agents import create_agent
agent = create_agent(
    model="deepseek-chat",
    tools=[search_tool]
)

from langchain_core.messages import HumanMessage
resp = agent.invoke({
    "messages": [
        HumanMessage("蒸蚌是什么梗？")
    ]
})

for message in resp["messages"]:
    message.pretty_print()
```
